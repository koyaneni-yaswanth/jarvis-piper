# JARVIS-PiPER :: Dataset Inspection & Metadata Provenance
This notebook demonstrates loading, inspecting, and validating recorded demonstration episodes.


```python
import os
import sys
import json
from pathlib import Path
import matplotlib.pyplot as plt

# Add python package to sys.path
sys.path.insert(0, str(Path.cwd().parent / 'python'))
from dataset.reader import EpisodeReader
from dataset.quality import QualityScorer
print('Imports successful!')
```

```python
raw_dir = Path('../dataset/raw')
episodes = sorted([p for p in raw_dir.iterdir() if p.is_dir() and (p / 'metadata.json').exists()])
print(f'Total episodes found: {len(episodes)}')
if episodes:
    reader = EpisodeReader(str(episodes[0]))
    print('Metadata:', json.dumps(reader.metadata, indent=2))
```

```python
if episodes:
    scorer = QualityScorer()
    audit = scorer.audit_episode(str(episodes[0]))
    print('Quality Audit:', json.dumps(audit, indent=2))
```
