# JARVIS-PiPER :: Multimodal Synchronization & Jitter Analysis
Quantifies temporal alignment, frame delta distributions, and dropped packets.


```python
import json
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / 'python'))
from evaluation.metrics import compute_sync_jitter
print('Imports successful!')
```

```python
raw_dir = Path('../dataset/raw')
episodes = sorted([p for p in raw_dir.iterdir() if p.is_dir() and (p / 'trajectory.json').exists()])
if episodes:
    with open(episodes[0] / 'trajectory.json') as f:
        data = json.load(f)
    ts = data.get('timestamps', [])
    stats = compute_sync_jitter(ts, expected_hz=50.0)
    print('Synchronization Jitter Stats:', json.dumps(stats, indent=2))
```

```python
if 'ts' in locals() and len(ts) > 1:
    diffs_ms = np.diff(ts) * 1000.0
    plt.figure(figsize=(8, 4))
    plt.hist(diffs_ms, bins=30, color='teal', edgecolor='black')
    plt.title('Inter-Frame Time Delta Distribution (Expected: 20 ms)')
    plt.xlabel('Delta t (ms)')
    plt.ylabel('Frame Count')
    plt.grid(True)
    plt.show()
```
