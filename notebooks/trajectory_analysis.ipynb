# JARVIS-PiPER :: Robot Trajectory & Kinematics Analysis
Analyzes joint position profiles, velocities, accelerations, and jerk across teleoperation demonstrations.


```python
import json
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / 'python'))
from evaluation.metrics import compute_trajectory_jerk
print('Imports successful!')
```

```python
raw_dir = Path('../dataset/raw')
episodes = sorted([p for p in raw_dir.iterdir() if p.is_dir() and (p / 'trajectory.json').exists()])
if episodes:
    with open(episodes[0] / 'trajectory.json') as f:
        data = json.load(f)
    joints = np.array(data.get('joint_positions', []))
    print(f'Joint trajectory shape: {joints.shape}')
    if len(joints) > 5:
        jerk = compute_trajectory_jerk(joints)
        print(f'Trajectory RMS Jerk: {jerk:.4f}')
```

```python
if 'joints' in locals() and len(joints) > 0:
    plt.figure(figsize=(10, 4))
    for j in range(min(6, joints.shape[1])):
        plt.plot(joints[:, j], label=f'Joint {j+1}')
    plt.title('PiPER 6-DOF Joint Positions over Time')
    plt.xlabel('Timestep')
    plt.ylabel('Angle (rad)')
    plt.legend()
    plt.grid(True)
    plt.show()
```
